# Sensor demo (synthetic)
Every row is generated; nothing here is real data.

```python
readings = [{"sensor": s, "step": i, "value": (i * 7 + s) % 11} for s in range(3) for i in range(8)]
print(len(readings), "synthetic rows")
```

```python
%%sql
SELECT sensor, avg(value)::DOUBLE AS mean_value
FROM readings
GROUP BY sensor
ORDER BY sensor
```
